# 01 — Validate the DAR dataset

Checks the expected columns, sample size, duplicate participant IDs, and missing values. Run from the repository root.

```python
from pathlib import Path
import pandas as pd

DATA = Path('data/dar_naturalized_data_70.csv')
expected = [
    'Participant', 'Pre_Density', 'Post_Density', 'Pre_Length', 'Post_Length',
    'Pre_Rejection', 'Post_Rejection', 'Pre_Stance', 'Post_Stance',
    'Pre_Lexical', 'Post_Lexical', 'Pre_Consistency', 'Post_Consistency',
    'Pre_Critical', 'Post_Critical'
]
df = pd.read_csv(DATA)
print('Rows:', len(df))
print('Missing expected columns:', sorted(set(expected) - set(df.columns)))
print('Duplicate participant IDs:', int(df['Participant'].duplicated().sum()))
display(df.isna().sum().rename('Missing values').to_frame())
assert len(df) == 70, f'Expected N=70; found {len(df)}'
assert set(expected).issubset(df.columns), 'Required columns are missing'
assert not df['Participant'].duplicated().any(), 'Participant IDs are duplicated'
print('Validation passed.')
```
