# 09 — Missingness and range audit

Summarizes missingness and observed ranges. This is a flagging tool, not an automatic validity decision.

```python
from pathlib import Path
import pandas as pd

DATA = Path('data/dar_naturalized_data_70.csv')
OUT = Path('outputs'); OUT.mkdir(exist_ok=True)
df = pd.read_csv(DATA)
numeric = df.select_dtypes(include='number')
audit = pd.DataFrame({'missing_n': numeric.isna().sum(), 'missing_pct': numeric.isna().mean()*100,
                      'minimum': numeric.min(), 'maximum': numeric.max(),
                      'mean': numeric.mean(), 'sd': numeric.std()})
audit.to_csv(OUT/'range_missingness_audit.csv')
display(audit.round(3))
```
