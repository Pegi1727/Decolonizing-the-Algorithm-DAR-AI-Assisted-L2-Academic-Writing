# 04 — Effect sizes

Uses the convention `r = |Z| / sqrt(N_total_pairs)`. State the convention explicitly in the manuscript; other conventions may use non-zero paired differences.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon

DATA = Path('data/dar_naturalized_data_70.csv')
OUT = Path('outputs'); OUT.mkdir(exist_ok=True)
measures = {
    'Prompting density': ('Pre_Density', 'Post_Density'),
    'Mean prompt length': ('Pre_Length', 'Post_Length'),
    'AI-suggestion rejection rate': ('Pre_Rejection', 'Post_Rejection'),
    'Stance marking': ('Pre_Stance', 'Post_Stance'),
    'Lexical agency': ('Pre_Lexical', 'Post_Lexical'),
    'Voice consistency': ('Pre_Consistency', 'Post_Consistency'),
    'Critical dialogue': ('Pre_Critical', 'Post_Critical'),
}
df = pd.read_csv(DATA)
rows = []
for name, (pre, post) in measures.items():
    pair = df[[pre, post]].dropna()
    test = wilcoxon(pair[post], pair[pre], method='approx')
    z = abs(float(test.zstatistic))
    rows.append({'Measure': name, 'N_total_pairs': len(pair), 'Z_abs_approx': z,
                 'r_Z_over_sqrt_N': z / np.sqrt(len(pair)), 'p_value': float(test.pvalue)})
result = pd.DataFrame(rows)
result.to_csv(OUT / 'effect_sizes.csv', index=False)
display(result.round(4))
print('Effect-size convention: r = |Z| / sqrt(N_total_pairs).')
```
