# 08 — Participant-level paired changes

Plots individual pre/post trajectories. Use only in a way consistent with participant confidentiality and data-sharing permissions.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA = Path('data/dar_naturalized_data_70.csv')
FIG = Path('figures'); FIG.mkdir(exist_ok=True)
measures = {
    'Density': ('Pre_Density', 'Post_Density'), 'Length': ('Pre_Length', 'Post_Length'),
    'Rejection': ('Pre_Rejection', 'Post_Rejection'), 'Stance': ('Pre_Stance', 'Post_Stance'),
    'Lexical agency': ('Pre_Lexical', 'Post_Lexical'),
    'Voice consistency': ('Pre_Consistency', 'Post_Consistency'),
    'Critical dialogue': ('Pre_Critical', 'Post_Critical'),
}
df = pd.read_csv(DATA)
fig, axes = plt.subplots(4, 2, figsize=(10, 13)); axes = axes.flatten()
for ax, (name, (pre, post)) in zip(axes, measures.items()):
    pair = df[[pre, post]].dropna()
    for _, row in pair.iterrows():
        ax.plot([0, 1], [row[pre], row[post]], alpha=0.25, linewidth=0.7)
    ax.set_xticks([0, 1]); ax.set_xticklabels(['Pre', 'Post'])
    ax.set_title(name); ax.set_ylabel('Observed value')
axes[-1].axis('off'); fig.tight_layout()
fig.savefig(FIG/'figure_paired_changes.png', dpi=300, bbox_inches='tight')
fig.savefig(FIG/'figure_paired_changes.pdf', bbox_inches='tight')
plt.show()
```
