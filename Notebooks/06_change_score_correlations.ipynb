# 06 — Exploratory correlations among change scores

Computes Spearman correlations among pre–post change scores. Treat these as exploratory associations, not evidence of causality.

```python
from pathlib import Path
import pandas as pd

DATA = Path('data/dar_naturalized_data_70.csv')
OUT = Path('outputs'); OUT.mkdir(exist_ok=True)
measures = {
    'Density_change': ('Pre_Density', 'Post_Density'),
    'Length_change': ('Pre_Length', 'Post_Length'),
    'Rejection_change': ('Pre_Rejection', 'Post_Rejection'),
    'Stance_change': ('Pre_Stance', 'Post_Stance'),
    'Lexical_change': ('Pre_Lexical', 'Post_Lexical'),
    'Consistency_change': ('Pre_Consistency', 'Post_Consistency'),
    'Critical_change': ('Pre_Critical', 'Post_Critical'),
}
df = pd.read_csv(DATA)
changes = pd.DataFrame({name: df[post] - df[pre] for name, (pre, post) in measures.items()})
corr = changes.corr(method='spearman')
corr.to_csv(OUT / 'change_score_spearman_correlations.csv')
display(corr.round(3))
```
